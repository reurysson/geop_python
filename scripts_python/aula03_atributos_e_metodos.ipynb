### Instrução `import`
Em um script Python, podemos importar blocos de códigos criados anteriormente, assim como fazemos com os pacotes nativos ou de terceiros. 

Mas antes de criar nosso primeiro módulo, precisamos entender que em Python tudo é objeto e que objetos têm tanto atributos quanto métodos. 

Módulos Python podem conter **atributos** e **funções**.
- Atributos: **São valores** armazenados dentro de um objeto. **Representam as características ou propriedades de um objeto ou módulo**.
- Funções: **São blocos de código** definidos dentro de um módulo que realizam uma tarefa específica. **Representam comportamentos ou ações que o objeto pode executar**.

Para criar um módulos, basta criar um arquivo `.py` contendo a funcionalidade desejada. Poderíamos criar um arquivo `.py` utilizando um editor de texto convencional, como o Bloco de Notas ou Notepad++. Usaremos o próprio Jupyter Lab para criar esse arquivo.

1. Clique em `File > New > Python File`.  
2. Uma nova aba será aberta com o nome `untitled.py`.  
3. Digite o código abaixo e salve o arquivo.
