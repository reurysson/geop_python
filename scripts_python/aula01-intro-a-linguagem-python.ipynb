## Expressões

Python segue a ordem de precedência conhecida como **PEMDAS** ao avaliar expressões matemáticas.  

**PEMDAS** é a sigla para:  
**P**arênteses → **E**xpoentes → **M**ultiplicação → **D**ivisão → **A**dição → **S**ubtração  

Essa ordem define como as operações são realizadas em uma expressão.
Quando duas operações estão no mesmo nível de precedência (como multiplicação e divisão), a avaliação ocorre **da esquerda para a direita**.

````python
6 + 6 ** 2 
````

````python
15 + 5 * 6
````

````python
15 + 5 * 6 / 2
````

````python
(12 + 5) * 6
````
