## Geração de números aleatórios

- Operação extremamente útil em análise de dados, especialmente em processo de obtenção de dados aleatórios oriundos de distribuições de probabilidade específicas. 
- Esse tema ficará mais claro após cursarem a disciplina de Estatística Aplicada.
- Focaremos apenas nas distribuições normal e normal padrão que são as mais comumente utilizadas em análise de dadps — pelo menos em nível introdutório. 

```python
# Vamos iniciar dando uma olhada na documentação do módulo random. 
# np.random?
```

```python
# rand: retorna números aleatórios em 0 e 1 provenientes de uma distribuição uniforme. 
amostra1 = np.random.rand(10)  # Cria uma array com 10 numeros aleatórios entre 0 e 1
amostra1
```

```python
# rand: retorna númerso aleatórios em 0 e 1. 
amostra2 = np.random.rand(3,5)  # Cria uma matriz com m linhas de n colunas preenchidas com valores aleatórios aleatórios entre 0 e 1
amostra2
```

```python
# Talvez deseje obter valore(s) aleatório(os) de números inteiros que ocorrem entre dois valores (low, high). 
# Para isso use o randint(low, high, n)

amostra3 = np.random.randint(10,100, 5)   # Gera 5 números aleatórios entre 10 e 100
print(amostra3)
```

```python
# Gera valor aleatório proveniente de uma distribuição normal com média µ e desvio pardrão σ
np.random.normal()
```

```python
# Gerando um array com n valores  aleatórios proveniente de uma distribuição normal com média µ (loc) e desvio pardrão σ (scale)
np.random.normal(loc=35, scale=3, size = 10)
```

```python
# Gerando uma matriz mn preenchida com valores aleatórios proveniente de uma distribuição normal com média µ (loc) e desvio pardrão σ (scale)
np.random.normal(loc=35, scale=3, size = (3,3))
```

```python
# Gerando um array com n valores aleatórios provenientes de uma distribuição normal padrão 
np.random.standard_normal(10)
```

```python
# Gerando uma matriz mn com valores aleatórios provenientes de uma distribuição normal padrão 
np.random.standard_normal(size=(4,4))
```

```python
# Em algumas situação é necessário controlar o processo de seleção aleatória, especialmente quando estamos interessados
# na replicabilidade do código. Para isso, definimos o estado inicial do gerador de números aleatórios.

semente = np.random.default_rng(seed=12345)
semente.standard_normal((2,3))
```

### Algumas funções universais 

```python
arr = np.arange(11)
arr
```

```python
# Raiz quadrada de cada elementos de uma array
np.sqrt(arr)
```

```python
# Quadrado de cada elemento de um array
np.square(arr)
```

```python
np.exp(arr)
```

```python
rng = np.random.default_rng(42)
x = rng.standard_normal(8)
y = rng.standard_normal(8)

print("x:",x)
print("y:", y)
print("-"*30)
print("Máximo:",np.maximum(x,y))  # Retorna valores máximo elemento a elemento;

print("-"*30)
print("Mínimo:",np.minimum(x,y)) # Retorna valores mínimos elemento a elemento;
```

```python
# Função absoluto
x_abs = np.abs(x)
print(x_abs)


# Transformação logaritmica
x_log = np.log10(x_abs)
print(x_log)
```

### Métodos matemáticos e estatísticos básicos

```python
np.random.seed(42)
arr = np.random.standard_normal((5, 4))
arr
```

```python
# Estatísticas da matriz

print("Soma:", arr.sum())
print("Mínimo:", arr.min())
print("Média:", arr.mean())
print("Desvio padrão:", arr.std())
print("Máximo:", arr.max())
```

```python
# Soma dos elementos de cada linhas

arr_sum_linha = arr.sum(axis=0)
print(arr_sum_linha)
```

```python
# Soma dos elementos de cada coluna 

arr.sum(axis=1).reshape(-1, 1)
```
