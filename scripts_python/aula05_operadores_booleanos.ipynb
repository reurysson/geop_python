## Booleano

Em Python, os dois valores booleanos são:
- `True` (Verdadeiro)
- `False` (False)

Esses valores são do tipo `bool` e são frequentemente usados em **comparações**, **condições** e **estruturas de decisão**.


```python
# True e False não são estrings, são valores especials do tipo bool.
type(True)
```

```python
print(int(True))
print(int(False))
```

## Operadores Lógicos

Em Python, os **operadores lógicos** são usados para combinar expressões booleanas. Eles retornam `True` ou `False` com base nas condições avaliadas.

Os principais operadores lógicos são: 
- `and`(&): retorna `True` apenas se **ambas** as expressões forem verdadeiras.
- `or`(|): retorna `True` se **pelo menos uma** das condições for verdadeira
- `not`: inverte o valor lógico de uma expressão. Se a expressão for `True`, ela se torna `False`, e vice-versa.

```python
x = 8

x > 0 and x <= 10  # Retorna True se ambas as condições forem verdadeiras
```

```python
x < 0 or x <= 10  # Retorna True se uma ou ambas as condições forem verdadeiras
```

```python
x < 0 or x >= 10  # Retorna True se uma ou ambas as condições forem verdadeiras
```

```python
x > 5
```

```python
not x > 5
```
