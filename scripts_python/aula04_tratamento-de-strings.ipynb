## Tratamento de Strings

Em análise de dados, é comum lidarmos com **dados categóricos**, como nomes, classificações e categorias textuais (nominais e ordinais).

Muitas vezes, é necessário **tratar e padronizar esses textos** antes de realizar análises.

A seguir, estão descritas as principais situações em que o tratamento de strings é necessário:

- Contar a quantidade de caracteres em um texto  
- Eliminar espaços desnecessários  
- Colocar o texto todo em **maiúsculas** ou **minúsculas**  
- Separar partes de palavras ou frases (por exemplo, dividir nomes e sobrenomes)
- Substituir ou remover palavras/padrões específicos  

O tratamento adequado de strings evita erros em agrupamentos, filtragens e análises estatísticas.


### Escrita de uma string

```python
# Strings são escritas entre aspas simples ou duplas (recomendado)

curso = "Tecnologia em Geoprocessamento"
disciplina = 'Introdução à Linguagem Python para Análise de Dados'
```

```python
print(type(curso))
print(type(disciplina))
```

```python
# Para escrita em multiplas linhas ou com quebra de linhas, usa-se aspas triplas 

apresentacao = """
O curso superior e tecnologia em Geoprocessamento é ofertado no 
campus Teresina Central do IFPI desde o ano 2000. Hoje, o IFPI 
conta também com um especialização em Geoprocessamento e uma mestrado 
em Análise e Planejamento Espacial, além do curso técnico em Agrimensura 
integrado ao ensino médio. 
"""
```

```python
apresentacao
```

```python
print(apresentacao)
```

**Atenção**

- Observe os caracteres "\n" na string exibida. Eles executam a função de quebra de linha, mas não aparecem no print()
- Para obtermos um texto sem quebra de linhas, devemos usar um barra invertida (\) ao final de nada linha.


```python
# Para evitar quebras de linhas usa-se uma barra invertida "\" 

apresentacao = """
O curso superior e tecnologia em Geoprocessamento é ofertado no \
campus Teresina Central do IFPI desde o ano 2000. Hoje, o IFPI \
conta também com um especialização em Geoprocessamento e uma mestrado \
em Análise e Planejamento Espacial, além do curso técnico em Agrimensura \
integrado ao ensino médio. 
"""
print(apresentacao)
```

### Contagem de caracteres

```python
# Podemos contar quantas vezes "\n" aparece, ou seja, quantas quebras de linhas há no texto.

apresentacao.count("\n")
```

```python
# Da mesmo forma podemos contar quantas vezes uma letras ou palavra aparece no texto.

apresentacao.count("Geoprocessamento")
```

```python
# Podemos facilmente contar quantos caracteres tem numa palavra ou frase usando a funsão len() de "length", 
# que retorcomprimento da cadeia de caracteres. Observe que espaços entram na contagem. 

print(curso)
len(curso)
```

### Segmentação de uma string

```python
# Podemos separar as palavras de uma frase usando o método .split(). 
# O resultado é uma lista com todas as palavras. 

curso.split()  # Como não foi passado nenhum argunmento para função, ela separou as palavras com base nos espacos.
```

```python
# Podemos separar todos os caracteres de um string e colocá-los em uma lista

print(list(curso))
```

```python
# Acessando apenas a palavra "Tecnologia". A notação [:] é usada para definir o início e o fim do fatiamento (slincig) da string.
# No exemplo abaixo usamos [:10], que retorna os caracteres que ocupam a posição 0 à 9.

curso[:10]  # Isso é equiivalente a curso [0:10]
```

```python
# Podemos acessar qualquer letra ou sequêcnia de letras de uma string através no seu índice. 
# Lembre-se sempre que a indexação em Python começa em "0". 

# Acessando a primeira letra da variável curso

print(curso[0])
```

```python
# Separando apenas a palavra "Geoprocessamento" 

# Modo 1:
print(f"Modo 1: {curso[14:]}")  # Isso é equivalente a curso[14:30]

# Modo 1:
print(f"Modo 2: {curso[-16:]}")  # Isso é equivalente a curso[-16:30]
```

### Pertinência (`in`) e Impertinência (`not in`)

```python
"Geo" in curso
```

```python
"logia" in curso
```

```python
"remoto" in curso
```

```python
"remoto" not in curso
```

```python
"Tec" not in curso
```

```python
# Para progurar um determinada palavra ou trecho de palavra pu frase, podemos usar o método .find()

curso.find("Tec")  # O número resultante é o índice associado à primeira lentra da texto pesquisado. 
```

```python
curso.find("Geo")
```

```python
curso.find("remoto") # O valor -1 ocorre quando o texto não é encontrado.
```

### Formatação de uma string

```python
# Transformando uma string em letras maiúsculas

curso.upper()
```

```python
# Transformando uma string em letras minúsculas

curso.lower()
```

```python
# Caso uma string possua espaços vazios antes ou depois, podemos removê-los com a função .strip()

tema = " formatação de strings "
print(len(tema))

tema_cor = tema.strip()
print(len(tema_cor))
```

### Alteração de uma string

```python
# Strings são imutáveis. Elas não podem ser alteradas. Observe o erro proposital na palavra "Geoprocessamento".
# O erro tá na letra que ocupa a posição 6 da string

a = "Geoproxessamento"
```

```python
a[6]
```

```python
# Observe o erro ao tentar substituir "f" por "g"

a[6] = "c"
```

```python
# Para modificar um string é necessário usar uma função que crie uma nova string. Usaremos o método "replace"

b = a.replace("Geoproxessamento", "Geoprocessamento")
b
```

### Concatenação de strings

```python
# Strings podem ser concatenadas (+)

termo1 = "Gep"
termo2 = "Python"

nome = termo1+termo2
nome
```

```python
edicao = "_2025.1"
```

```python
nome+edicao
```

### Repetição de strings

```python
# Repetição de strings
(nome + "-") * 20
```
