## Variáveis e Instruções
Uma **variável** é um nome ao qual atribuimos um valor. 

Você pode pensar nela como um **objeto** ou até uma **caixa** que armazena determinado valor enquanto o programa está sendo executado.
Para "colocar" algo dentro de um variável, usamos uma **instruções de atribuição**.

```
n = 42
```

No exemplo acima, lemos: "n recebe 42". A partir desse momento, n passa a armazenar o número 42 até que o programa seja encerrado, ou até um novo valor seja atribuído a n. Observe que o sinal `=` é um operador de atribuição, e não um operador de igualdade, que em Python é definido como `==`.


````python
n = 42
print(n)

n = 42*4
print(n)
````

````python
pi = 3.141592653589793
pi
````

Instruções básicas para nomes de variáveis:
- nomes de variáveis **não têm limites de tamanho**, mas mantenha o mais curto e autoexplicativo possível;
- podem conter letras ou números, desde que o número não apareça no início do nome;
- podem conter letras maiúsculas e minúsculas, mas por convenção da comunidade, **utiliza-se apenas letras minúsculas**;
- o único símbolo permido em nomes de variáveis é o `_` (*underscore*), que pode ser utilizado para separar palavras;
- não se usa números como nome de variáveis:
    - Tente você mesmo: `42 = 30`
- Python possui palavras reservadas que não podem ser utilizadas para nomear variáveis.

### Instrução `import`

Já usamos essa instrução antes. As instruções `import` é usada para **trazer recursos externos** para o seu programa Python. 

Esses recursos podem ser:
- Módulos **nativos** da linguagem (como `math`, `keyword`, `random`, `statistic` etc.)
- **Pacotes criados por terceiros**, instalados via `pip`
- **Scripts ou arquivos Python pelo usuário**, em outros arquivos `.py`


````python
# Importanto o módulo de matemática do Python
import math
````

````python
# Acessando o valor de pi 
math.pi
````

````python
# Podemos usar a "dir" para listar todas as funções de um módulo.
print(dir(math))
````

````python
math.pow(5, 2)
````

````python
# Obterndo a rais quadrada de um número
math.sqrt(81)
````

````python
#### Calculando o seno, cosseno e tangente de um ângulo dado em graus
ang = 38
seno = math.sin(ang)
cosseno = math.cos(ang)
tangente = math.tan(ang)

print(f'Seno: {seno:.3f}') 
print(f'Cosseno: {cosseno:.3f}')
print(f'Tangente: {tangente:.3f}')
````

````python
# Tranformando um ângulo dado graus para radiano
math.radians(38.5)
````

### Função `Print`

A função `print` é um recurso fundamental do Pyrhon. Ela permite **exibir informações na tela**, o que é essencial para entender os *outputs* (saídas) de expressões, funções e resultados de processos em geral.

Vamos explorar algumas configurações dessa função:

````python
print(seno)
````

````python
# Print básico

print("O seno de 38 é: ")
print(seno)
````

````python
# Print formatado

print(f"O seno de {ang}° é: {seno}") 
````

````python
# Definindo a quantidade de casas decimais do número de saída 

print(f"O seno de {ang}° é: {seno:.3f}") 
````

````python
# Print em duas linhas 

print(f"Resultados:\nSeno: {seno:.3f}\nCosseno: {cosseno:.3f}\nTangente: {round(tangente,4)}") 
````

````python
# Print em duas linhas 

print(f"Resultados:\nSeno: {seno:.3f}\nCosseno: {cosseno:.3f}\nTangente: {round(tangente,4)}") 
````

#### Print formatado
Uma situação especial ocorre ao utilizar um print formatado. Prints formatadas usam a sintaxe `print(f"")`. Podemos substituir as aspas duplas por aspas simples. Só não podemos misturá-las dessa forma: `print(f"Python')`

`print("Python")` e o mesmo que `print(f'Python')`

Sempre que a string contiver aspas simples internas, use aspas duplas externas — e vice-versa. Outra solução seria usar o caractere de escape "\\".

````python
print("Barra D'Alcântara")
````

````python
print('Barra D\'Alcântara')
````

### Argumentos
Funções podem receber **um ou mais argumentos**. 
- A função `int()`, que usamos para transforma uma string ou n° float em um n° inteiro, espera receber um único argumento.
- A função `math.pow()`espera receber dois argumentos: a base e do expoente.
- A função `print` conseque receber qualquer quantidade de argumentos separados por vírgula. 

Agumentos de um a função podem ser: múmeros, strigns e até mesmo outras funções.
