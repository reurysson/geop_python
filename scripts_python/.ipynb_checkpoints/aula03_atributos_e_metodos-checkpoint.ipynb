### Instrução `import`
Em um script Python, podemos importar blocos de códigos criados anteriormente, assim como fazemos com os pacotes nativos ou de terceiros. 

Mas antes de criar nosso primeiro módulo, precisamos entender que em Python tudo é objeto e os objetos têm tanto atributos atributos quanto métodos. 

Módulos Python podem conter **atributos** e **métodos**.
- Atributos: São valores armazenados dentro de um objeto. Representam as características ou propriedades de um objeto.
- Métodos: São funções associadas a um objeto. Representam comportamentos ou ações que o objeto pode executar.

Para isto, basta criar um arquivo `.py` contendo da funiconalidade desejada. 

1. Clique em `File > New > Python File`.  
2. Uma nova aba será aberta com o nome `untitled.py`.  
3. Digite o código abaixo e salve o arquivo.  
4. Em seguida, **renomeie o arquivo para** `meu_modulo.py`.

```
curso = "Tecnologia em Geoprocessamento"

disciplina = "Introdução à Linguagem Python para Análise de Dados"

modulo = "III"

```

O módulo criado contém agora três atributos.

````python
# Agora vamos importar o módulo recém criado
import meu_modulo
````

````python

print(f"Curso: {meu_modulo.curso}")
print(f"Disciplina: {meu_modulo.disciplina}")
print(f"Período: {meu_modulo.periodo}")
````

#### Agora vamos adicionar um método ao módulo recém criado. 

Abra o arquivo `meu_modulo.py`e digite o código abaixo:

```
def saudacao(nome):
    print(f"Seja bem vindo a disciplina de Python, {nome}!")  
```

````python
mensagem = meu_modulo.saudacao("Francisco")
````

#### Vamos adicionar ao módumo `meu_modulo.py`uma nova função que soma dois números. 

Adicone a função abaixo ao arquivo `meu_modulo.py`

```
def soma(n1, n2):
    return n1 + n2
```
Salve as alterações. Agora podemos usar a função de soma presente no módulo.

````python
meu_modulo.soma(45, 26)
````
