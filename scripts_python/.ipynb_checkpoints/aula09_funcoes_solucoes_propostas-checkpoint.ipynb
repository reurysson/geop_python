# Desafio 1

Nos exemplos anteriores, criamos três funções separadas para realizar cálculos relacionados à escala cartográfica. Agora, seu desafio é pensar em uma forma de unificar esses cálculos em uma única função.

A função deve ser capaz de receber quaisquer dois dos três valores a seguir:

- distância gráfica (em centímetros)

- distância real (em quilômetros)

- numerador da escala

Com base nos valores informados, a função deverá identificar automaticamente qual variável está faltando e calcular o valor correspondente.

Exemplos de uso esperados:
- Se o usuário informar a distância real e a distância gráfica, a função deve retornar o numerador da escala.

- Se o usuário informar a distância gráfica e o numerador da escala, a função deve retornar a distância real.

- Se o usuário informar a distância real e o numerador da escala, a função deve retornar a distância gráfica.

Orientações:
Instrua o usuário a fornecer os valores nas unidades apropriadas:

- distância gráfica em centímetros (cm)

- distância real em quilômetros (km)

- Faça uso de docstring para documentar a função.

- Use `0` como valor padrão para os argumentos, para identificar qual valor não foi informado.



```python
def calcula_escala(dist_grafica, dist_real, escala):
    """
    Calcula o valor que estiver faltando entre distância gráfica, distância real e escala cartográfica.

    Parâmetros:
    - dist_grafica: distância no mapa (em centímetros). Se não souber, informe 0.
    - dist_real: distância no terreno (em quilômetros). Se não souber, informe 0.
    - escala: numerador da escala (sem unidade). Se não souber, informe 0.

    Exemplo:
    - Se souber a distância gráfica e a distância real, a função calcula a escala.
    - Se souber a escala e a distância gráfica, a função calcula a distância real.
    - Se souber a escala e a distância real, a função calcula a distância gráfica.
    """

    if dist_grafica == 0 and dist_real !=0 and  escala !=0:
        # calcula distância gráfica
        dist_cm = dist_real * 100000
        dist_grafica = dist_cm / escala
        print(f"Distância gráfica: {dist_grafica:.2f} cm")

    elif dist_real == 0 and dist_grafica !=0 and escala !=0:
        # calcula distância real
        dist_cm = dist_grafica * escala
        dist_real = dist_cm / 100000
        print(f"Distância real: {dist_real:.2f} km")

    elif escala == 0 and dist_real !=0 and dist_grafica !=0:
        # calcula a escala
        dist_cm = dist_real * 100000
        escala = dist_cm / dist_grafica
        print(f"Escala: 1:{escala:.0f}")

    else:
        print("Informe dois valores diferentes de zero (O). Deixe o valor desconhecido como 0.")
```

# Desafio 2
Crie uma função chamada decimal_para_dms() que converta coordenadas geográficas em graus decimais (ex: -5.08972) para graus, minutos e segundos (DMS).

A função deve aceitar um par de valor decimal e o tipo ('lat' ou 'lon').

Identifique se a coordenada é N/S (latitude) ou E/W (longitude).

Retorne o valor como string formatada no formato 0°00'000"S.

```python
def decimal_dms(lat, lon):
    # LATITUDE
    if lat < 0:
        lat_h = "S"
    else:
        lat_h = "N"
    lat_abs = abs(lat)
    lat_g = int(lat_abs)
    lat_m_total = (lat_abs - lat_g) * 60
    lat_m = int(lat_m_total)
    lat_s = (lat_m_total - lat_m) * 60

    # LONGITUDE
    if lon < 0:
        lon_h = "O"
    else:
        lon_h = "L"
    lon_abs = abs(lon)
    lon_g = int(lon_abs)
    lon_m_total = (lon_abs - lon_g) * 60
    lon_m = int(lon_m_total)
    lon_s = (lon_m_total - lon_m) * 60

    # Exibição
    print(f"Latitude:  {lat_g}°{lat_m}'{lat_s:.2f}\" {lat_h}")
    print(f"Longitude: {lon_g}°{lon_m}'{lon_s:.2f}\" {lon_h}")

```

```python

lat = -5.2458
long = -42.4568

decimal_dms(lat, long)
```
