## Controle de fluxo

Instruções de controle de lfuxo permitem que o programa **tome decisões** e **repita ações**, comportando-se de maneira dinâmica com base nas condições especificadas pelo usuário.

Principais instruções de controle de fluxo:
- `if`, `elif` e `else`
- Loops `for`
- Loops `while`: 

### Instruções `if`, `elif` e `else`
Verifica se uma condiçõe é verdadeira. Caso seja executa o bloco de código seguinte.

Vejamos um exemplo:

Suponha que você queira fazer um programa que exiba na tela o status do aluno (aprovado, reprovado, prova final) a partir da média final de uma disciplina

```python
# Implementando uma instrução if 

media_final = 7.5

if media_final >= 7.0:
    print("Parabéns! Você foi aprovado!")  # Atenção para a identação
```

O programa acima funcionou bem. Mas o que acontece se a média final por menor que 7.0? Observe que devemos implementar mais uma solução para situações em que a média final seja menor que 7.0. Usaremos a intrução `elif`.

```python
# Agora precisamos implementar uma solução para quando a nota for menor que 7.0

media_final = 6.8

if media_final >= 7.0:
    print("Parabéns! Você foi aprovado!")
elif media_final < 7.0:
    print("Você ficou de prova final.")
```

O programa melhou um pouco, mas há um erro. Aluno com **média final menor que 4.0 fica reprovado direto**, sem direto a prova final. Então precisamos implementar mais uma solução. Usaremos uma combinação de **análise condicional** com a instrução `else`.

```python
# Agora precisamos implementar uma solução para quando a nota for menor que 7.0

media_final = 3.9

if media_final >= 7.0:
    print("Parabéns! Você foi aprovado!")
elif media_final < 7.0 and media_final >= 4.0:
    print("Você ficou de prova final.")
else:
    print("Você está reprovado.")
```

Em resumo:
- o `if` testa se a primeira condição é verdadeira. Caso seja, o bloco de código correspondente é executado e o programa não avalia as demais condições.
- Caso a primeira condição seja falsa, a segunda condição é testada com `elif`. Sendo verdadeira o programa executa o bloco de código correspondente e para.
- Se nenhuma condição for verdadeira, o `else` é executado como última alternativa. 

Numa análise condicional, como a do exemplo, podemos ter quantos `elif` forem necessários para combrir todas as condições possíveis. No entanto, só é permitido um único `if`e um único `else`por estrutura condicional.
