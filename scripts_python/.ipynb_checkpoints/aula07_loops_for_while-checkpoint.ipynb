## Loops for

Loops for são utilizados para iterar sobre item de qualquer sequência (listas, tuplas ou dicionários), na ordem em que aparecem na sequência.

Estrutura básica:
```
for valor in colecao:
    # faz algo com valor
```

````python
# Exemplo básico
palavras = ["Geoprocessamento", "Cartografia", "Sensoriamente Remoto"]

for palavra in palavras:
    print(palavra, len(palavra))
````

````python
# Na lista abaixo, temos inteiros, floats e strings. Vamos converter todos em inteiros
seq = [1, 2, "3", 4.0, 5]

# Vamos iniciar criando uma lista vazia
seq_int = []
for x in seq:               # Para cada "x" na lista "seq" ..
    x_int = int(x)          # a variável "x_int" vai receber "x" e converter para inteiro.
    seq_int.append(x_int)   # Em seguida, "x_int" é adionado à lista vazia "seq_int".
print(seq_int)
````

````python
# Um modo mais eficiente de fazer isso é usar list comprehension
seq = [1, 2, "3", 4.0, 5]

seq_int= [int(x) for x in seq]
seq_int
````

````python
# Vamos elevar cada número da lista anterior ao quadrado
seq_2 = []
for x in seq_int:
    x = x**2
    seq_2.append(x)
print("for normal:", seq_2)

# Modo list comprehension
seq_2 = [x**2 for x in seq_int]
print("list comprehension:", seq_2)
````

````python
# Adicionando uma constante a cada valor de uma sequência
seq_3 = []
for x in seq_int:
    x += 10        # a variável "x" recebe ela mesmo + 10
    seq_3.append(x)
print(seq_3)
````

````python
# Somando os valores de uma lista
seq_4 = [1, 2, 3, 4, 5, 6]

total = 0
for x in seq_4:
    total += x
print(total)
````

````python
# Somando apenas os valores numéricos de uma lista
seq_5 = [1, 2, "3", 4, "5", 6]
total = 0
for x in seq_5:
    if isinstance(x, str):   # Verifica de algum valor de entrada é uma string
        continue             # Ignora strings e continua a soma
    total += x 
print(total)
````

````python
# Somando os valores numéricos até um determinado valor
print(seq_4)
total = 0
soma = 0

for x in seq_4:
    if total >= 10:
        break        # Para a soma qual o total for maior ou igual a 10
    total += x
    soma += 1
    print(f"{soma}ª Soma: {total}")
    
print(10*"_")
print(total)
````

````python
# Calculando o valor do menor objeto representável em um mapa à partir de sua escala
escalas = [1000000, 500000, 250000, 100000, 75000, 50000, 25000, 10000]

dimensao_obj = []
for escala in escalas:
    erro = 0.0002 * escala
    dimensao_obj.append(int(erro))
    
print(dimensao_obj)
````

### Loops while
Loop while especifica uma condição e um bloco de código é executado até a condição ser avaliada com `False` ou o loop ser encerrado com a instrução `break`.


````python
i = 1

while i < 10:
    print(i)
    i += 1
````

````python
i = 1

while i < 10:
    print(i)
    if i == 3:
        break
    i += 1
````

````python
dados = [10, 20, 30, 40, 50]
total = 0
contador = 0

while contador < len(dados):
    total += dados[contador]
    contador += 1

media = total / contador
print("Média dos valores:", media)
````

````python
dados = [10, 15, 22, 28, 30, 18, 50]
indice = 0

while indice < len(dados):
    if dados[indice] > 25:
        print("Primeiro valor acima de 25 encontrado:", dados[indice])
        break  # Paramos a busca ao encontrar o primeiro valor
    indice += 1
````
