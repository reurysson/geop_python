#### Questão 01

Diversos estudos voltado para o planejamento territorial e ambiental necessitam de análise da declividade do terreno. Embora não haja um sistema oficial de classificação da declividade, a sistema apresentado pela Embrapa é bastante utilizado em trabalhos acadêmicos e profissionais. Segundo Santos (2018), são reconhecidas as seguintes classes de relevo:  
- Plano: 0% a 3%.
- Suave ondulado: 3% a 8%.
- Ondulado: 8% a 20%.
- Forte ondulado:20% a 45%.
- Montanhoso: 45% a 75%.
- Escarpado: maior que 75%
  
**A partir das classes de declividade acima, crie um programa que exiba a classe de declividade a partir de valor de entrada.**


Referência

SANTOS, H. G. dos. **Sistema brasileiro de classificação de solos**. 5 rev. e ampl.ed. Brasília, DF: Embrapa, 2018. 

```python
# Solução 1

declive = 37

if declive <= 3:
    print("Relevo plano")
elif declive > 3 and declive <= 8:
    print("Relevo suave ondulado")
elif declive > 8 and declive <= 20:
    print("Relevo ondulado")
elif declive > 20 and declive <= 45:
    print("Relevo forte ondulado")
elif declive > 45 and declive <= 75:
    print("Relevo montanhoso")    
else:
    print("Relevo escarpado")
```

```python
# Solução 2 - Nesta solução o usuário deverá informar a declividade em uma caixa de entrada que será mostrada ao executar o script.

declive = float(input("Insira um valor de declividade: "))

if declive <= 3:
    print("Relevo plano")
elif declive > 3 and declive <= 8:
    print("Relevo suave ondulado")
elif declive > 8 and declive <= 20:
    print("Relevo ondulado")
elif declive > 20 and declive <= 45:
    print("Relevo forte ondulado")
elif declive > 45 and declive <= 75:
    print("Relevo montanhoso")    
else:
    print("Relevo escarpado")
```

#### Questão 02

Sabe-se que o Índice de Vegetação por Diferença Normalizada (NDVI, sigla em inglês) varia na escala de -1 a 1. Empiricamente, podemos associar os seguintes intervalos de NDVI à classes de cobertura da terra: 

- água: ndvi < 0
- solo exposto: 0 <= ndvi < 0.1  
- Vegetação rala: 0.1 <= ndvi < 0.2
- Vegetação esparça: 0.2 <= ndvi < 0.4
- Vegetação transicional: 0.4 <= ndvi < 0.6
- vegetação densa: 0.6 <= ndvi <=1

**Com base nisso, crie um programa que receba um valor de ndvi fornecido pelo usuário e exiba 

```python
# Solução proposta

ndvi = float(input("Digite o valor do NDVI (entre -1 e 1): "))

if ndvi < -1 or ndvi > 1:
    print("Valor inválido! NDVI deve estar entre -1 e 1.")
elif ndvi < 0:
    print("Classe: Água")
elif ndvi < 0.1:
    print("Classe: Solo exposto")
elif ndvi < 0.2:
    print("Classe: Vegetação rala")
elif ndvi < 0.4:
    print("Classe: Vegetação esparsa")
elif ndvi < 0.6:
    print("Classe: Vegetação transicional")
else:
    print("Classe: Vegetação densa")
```
