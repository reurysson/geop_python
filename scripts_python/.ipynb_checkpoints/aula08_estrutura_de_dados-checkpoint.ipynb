### Listas

Listas é uma estrutuda de dados bastante versáril, elas podem armazenar qualquer tipo dado (string, números, booleano, etc) e até mesmo outros objetos e outras estrutura de dados. **Listas são mutáveis**, o que significa que podem ser alteradas a qualquer momento. 

Listas são definidas pelo uso de "[]". A lista abaixo armazema números (int e float) e palavras:
```
minha_lista = [1, 2.5, "Teresina", "07-04-2025"]
```
Listas podem ser criadas com a função `list()`. Vejamos alguns exemplos:


````python
# Lista de cidades
cidades = ["Teresina", "Altos", "José de Freitas"]
````

````python
# População das cidades da lista_cidades
pop_cidades = [866300, 47453, 42559]
pop_cidades
````

````python
# Podemos acessar o conteúdo de cada lista por meio do índice
print(f"{cidades[0]}: {pop_cidades[0]}")
print(f"{cidades[2]}: {pop_cidades[2]}")
````

````python
# Listas aceitam fateamento tal qual strings
cidades[:2]
````

````python
cidades[-1]
````

#### Adicionando elementos a uma lista

````python
cidades.append("Nazária")
pop_cidades.append(10262)
print(cidades)
print(pop_cidades)
````

#### Ordenando uma lista

````python
cidades.sort()
cidades
````

**Atenção!**

Ter ordenado a lista de cidade em ordem crescente criou um problema. Agora ficou difícil associar o nome da cidade ao dado de população. Isso foi proposital. Veremos com resolver isso mais adiante usando outra estrutura de dados.

#### Adicionando elementos numa posição específica


````python
# Vamos adionar a cidade Demerval Lobão mantendo a ordenação. Para isso, 
# precisamos inseror Demerval Lobão no índice 1. Lembre-se, a indexação em Python começa em zero

cidades.insert(1, "Demerval Lobão")  
print(cidades)
````

#### Fazendo um cópia de uma lista

````python
cidades_copia = cidades.copy()
````

#### Obtendo o comprimento de uma lista

````python
len(cidades)
````

#### Removendo o último elemento de uma lista

````python
cidades.pop()
cidades
````

#### Removendo um elementos específico de uma lista


````python
cidades.remove("Demerval Lobão")
cidades
````

#### Comparando duas listas

````python
cidades ==  cidades_copia
````

````python
print(cidades)
print(cidades_copia)
````

#### Limpando uma lista

````python
pop_cidades.clear()
pop_cidades
````

#### Criando lista com a função `range`


````python
# Criando uma sequencia de 1 a 10
seq1 = list(range(11))
seq1
````

````python
# Criando um lista com valores de 10 a 100
seq2 = list(range(10, 101))
print(seq2)
````

````python
# Criando um lista com valores de 0 a 100 com step de 5
seq2 = list(range(0, 101, 5))
print(seq2)
````

### Tuplas

Tuplas são parecidas com listas, mas ***são imutáveis**. Use tuplas, ao invés de listas, quando quisar garantir que os dados armazenados não sejam modificados.

Tuplas são definidas pelo uso de "()". A tupla abaixo armazema números (int e float) e strings:
```
minha_tupla = (1, 2.5, "Teresina", "07-04-2025")
```
Listas podem ser criadas com a função `tuple()`. 

As duas únicas funções associadas a tuplas são:
- `index`: retorna o índice de um elemento
- `count`: conta as ocorrêcias de determinado valor

````python
minha_tupla = (1, 2.5, "Teresina", "07-04-2025", "Teresina")
````

````python
minha_tupla.index(2.5)
````

````python
minha_tupla.count("Teresina")
````

**Por não permiter alterações, tuplas são úteis para armazenar coodenadas geográficas.**

````python
coord_teresina = ((-42.81109, -5.0904),(-41.7536, -2.9096))
coord_teresina[0]
````

**Uma variável que recebe um conjunto de valores sem especificação da estrutura de dados, o armazenará como tupla**

````python
coord  = -42.81109, -5.0904
print(coord)  #Observe os parênteses
type(coord)
````

````python
# Podemos transformar qualquer sequência em tuplas
# Convertendo uma lista em tupla
a = ["x", "y", "z"]
print(type(a))

b = tuple(a)
print(type(b))

# Convertendo uma tupla em lista
c = list(b)
print(type(c))

# Convertendo uma string em tupla
d = "Python"

e = tuple(d)
print(e)
````

#### Desempacotamento de tuplas

Um operação muito utilizada em análise de dados é o desempacotamento de tuplas. Muito frequentemente, os resultados de testes estatísticos são apresentados em formatos de tuplas. Quando estamos interessados em um resultados específico, podemos acessá-lo utilizando desempacotamento de tuplas. 

````python
dados = 9.544, 0.000458, 3.457

a, b, c = dados
print(a)
print(b)
print(c)
````

````python
# Um exemplo com um teste estatístico (teste de normalidade)
import scipy
idade = 25, 27,22, 26, 25, 24, 29, 31, 28

teste_norn = scipy.stats.shapiro(idade)
teste_norn
````

````python
test, p_valor = teste_norn
p_valor
````

````python
_, p_valor = teste_norn  # O undescore ignora tudo que vem antes do último resultado disponível 
p_valor
````

### Dicionários (`dict`)

Dicionários são de extrema importância em análise de dados. Eles armazenam uma coleção de pares chave-valor, em que chave e valor são objetos Python. As chaves devem ser únicas e imutáveis (como strings, números ou tuplas), enquanto os valores podem assumir qualquer tipo de dado ou estrutura de objetos Python (números, strings, lista, tuplas e dicionários). Podemos inserir, excluir, modificar valores de um dicionário a qualquer momento.

Dicionários são definidos por um par de `{}`. Dois pontos (`:`) separam a chave do valor de um dicionário.

Estrutura básica: `{chave: valor}`.


````python
# Dicionário básico
dic1 = {"Nome": "Francisco da Silva", "idade": 23, "peso": 84.5}
dic2 = {7: "é um número inteiro", 4.5: 'é um número decimal'}
````

````python
# Criando um dicionário com dados de população de algumas capitais do nordeste brasileiro

capitais = {"Teresina": 866300, "Fortaleza": 2428678, "São Luis": 1037775}
capitais
````

````python
# Dionários podem receber tuplas, listas e até mesmo outro dicionário como valor
# Ex. população e área (km²) de algumas capitais
capitais = {"Teresina": [866300, 1391293], "Fortaleza": [2428678, 312353], "São Luís": [1037775, 583063]}
````

````python
# Acessando os valores de uma chave específica em um dicionário
capitais["Teresina"]
````
