### Listas

Listas é uma estrutuda de dados bastante versáril, elas podem armazenar qualquer tipo dado (string, números, booleano, etc) e até mesmo outros objetos e outras estrutura de dados. **Listas são mutáveis**, o que significa que podem ser alteradas a qualquer momento. 

Listas são definidas pelo uso de "[]". A lista abaixo armazema números (int e float) e palavras:
```
minha_lista = [1, 2.5, "Teresina", "07-04-2025"]
```
Listas podem ser criadas com a função `list()`. Vejamos alguns exemplos:


````python
# Lista de cidades
cidades = ["Teresina", "Altos", "José de Freitas"]
````

````python
# População das cidades da lista_cidades
pop_cidades = [866300, 47453, 42559]
pop_cidades
````

````python
# Podemos acessar o conteúdo de cada lista por meio do índice
print(f"{cidades[0]}: {pop_cidades[0]}")
print(f"{cidades[2]}: {pop_cidades[2]}")
````

#### Adicionando elementos a uma lista

````python
cidades.append("Nazária")
pop_cidades.append(10262)
print(cidades)
print(pop_cidades)
````

#### Ordenando uma lista

````python
cidades.sort()
cidades
````

**Atenção!**

Ter ordenado a lista de cidade em ordem crescente criou um problema. Agora ficou difícil associar o nome da cidade ao dado de população. Isso foi proposital. Veremos com resolver isso mais adiante usando outra estrutura de dados.

#### Adicionando elementos numa posição específica


````python
# Vamos adionar a cidade Demerval Lobão mantendo a ordenação. Para isso, 
# precisamos inseror Demerval Lobão no índice 1. Lembre-se, a indexação em Python começa em zero

cidades.insert(1, "Demerval Lobão")  
print(cidades)
````

#### Fazendo um cópia de uma lista

````python
cidades_copia = cidades.copy()
````

#### Contando elementos de uma lista

````python
len(cidades)
````

#### Removendo o último elemento de uma lista

````python
cidades.pop()
cidades
````

#### Removendo um elementos específico de uma lista


````python
cidades.remove("Demerval Lobão")
cidades
````

#### Comparando duas listas

````python
cidades ==  cidades_copia
````

````python
print(cidades)
print(cidades_copia)
````

#### Limpando uma lista

````python
pop_cidades.clear()
pop_cidades
````

#### Criando lista com a função `range`


````python
# Criando uma sequencia de 1 a 10
seq1 = list(range(11))
seq1
````

````python
# Criando um lista com valores de 10 a 100
seq2 = list(range(10, 101))
print(seq2)
````

````python
# Criando um lista com valores de 0 a 100 com step de 5
seq2 = list(range(0, 101, 5))
print(seq2)
````

### Tuplas

Tuplas são parecidas com listas, mas ***são imutáveis**. Use tuplas, ao invés de listas, quando quisar garantir que os dados armazenados não sejam modificados.

Tuplas são definidas pelo uso de "()". A tupla abaixo armazema números (int e float) e palavras:
```
minha_tupla = (1, 2.5, "Teresina", "07-04-2025")
```
Listas podem ser criadas com a função `tuple()`. 

As duas únicas funções associadas a tuplas são:
- `index`: retorna o índice de um elemento
- `count`: conta as ocorrêcias de determinado valor

````python
minha_tupla = (1, 2.5, "Teresina", "07-04-2025", "Teresina")
````

````python
minha_tupla.index(2.5)
````

````python
minha_tupla.count("Teresina")
````

**Por não permiter alterações, tuplas são úteis para armazenar coodenadas geográficas.**

````python
coord_teresina = ((-42.81109, -5.0904),(-41.7536, -2.9096))
coord_teresina[0]
````

**Uma variável que recebe uma lista de valores sem especificação da estrutura de dados, o armazenará como tupla**

````python
dados  = 1, 5, 7
print(dados)  #Observe os parênteses
type(dados)
````
