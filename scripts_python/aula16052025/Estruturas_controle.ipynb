## Controle de fluxo

Instruções de controle de fluxo permitem que o programa **tome decisões** e **repita ações**, comportando-se de maneira dinâmica com base nas condições especificadas pelo usuário.

Principais instruções de controle de fluxo:
- `if`, `elif` e `else`
- Loops `for`
- Loops `while`: 

### Instruções `if`, `elif` e `else`
Verifica se uma condiçõe é verdadeira. Caso seja, executa o bloco de código seguinte.

Vejamos um exemplo:

Suponha que você queira fazer um programa que exiba na tela o status do aluno (aprovado, reprovado, prova final) a partir da média final de uma disciplina



```python
media = 9.8

if media >= 7.0:
    print("Parabéns! Você foi aprovado!")
elif media < 7.0 and media >= 4.0:
    print("Você ficou de prova final")
else:
    print("Você foi reprovado")
```
