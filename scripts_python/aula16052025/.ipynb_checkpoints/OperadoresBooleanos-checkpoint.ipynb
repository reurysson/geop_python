## Booleano

Em Python, os dois valores booleanos são:
- `True` (Verdadeiro)
- `False` (False)

Esses valores são do tipo `bool` e são frequentemente usados em **comparações**, **condições** e **estruturas de decisão**.

```python
type(True)
```

```python
print(int(True))
print(int(False))
```

```python
a = "p"
b = "P"
a == b
```

```python
a != b
```

```python
a is b
```

```python
a is not b
```

```python
a.isnumeric()
```

```python
x = 5
y = 7

x > y
```

```python
x < y
```

```python
9 == 3**2
```

```python
nome = "python"
len(nome) == 6
```
