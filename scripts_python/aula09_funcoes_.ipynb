## Funções
- As funções são o principal e mais importante método de organização e reutilização de código em Python. 
- Se precisar repertir uma operação várias vezes no mesmo código ou usar o mesmo procedimento em diferentes códigos, criar funções é a solução mais eficiente. 
- O uso de funções também deixa o código mais legível e fácil de dar manutenção.
- Funções são definidas pela palavra-chave `def`. Um função contém um bloco de código e opcionalmente pode usar a palavra-chave `return`.
- A primeira linha da definição de função chama-se cabeçalho; o resto é chamado de corpo. O cabeçalho precisa terminar em dois pontos e o corpo precisa ser endentado. O corpo pode conter qualquer número de instruções.

Estrutura típica de um função:
```
def area_quadrado(a, b):   # Cabeçalho
    area = a x b           # Corpo contendo uma variável e uma função print
    print(f"Área: {area}")

```
Funções podem ou não receber argumentos. Os parenteses vazios da função abaixo indica que ela não recebe nenhum argumento:

````python
def pi():
    return 3.14159265359

pi()
````

Já a próxima função recebe dois valores como argumentos e retorna sua soma. 

```
def func_soma(x, y):
    return x + y
```

````python
# Implementando a função de exemplo
def func_soma(x, y):
    return x + y
````

````python
# Função criada, vamos utilizá-la para somar dois valores. 
# Observe que no exemplo estamos passondo diretamente os valores com argumentos da função

func_soma(5, 8)
````

````python
# Mais um exemplo. Neste caso, temos duas variáveis previamente definidas que armazenam determinados valores que serão utilizados na soma. 

n1, n2 = 11, 17

func_soma(n1, n2)
````

## Argumentos de uma função:

Cada função pode ter argumentos **posicionais** e **argumentos nomeados**. 

Argumentos nomeados são utilizados para especificar valores padrão o argumentos opcionais. 

No exemplo abaixo, `x` e `y` são argumentos posicionais, enquanto `z` é um argumento nomeado:

```
def multiply_sum(x, y, z = 1.5)
    return x * (x + y)
```

````python
def my_function2(x, y, z = 1.5):
    if z > 1:
        return z * (x + y)
    else:
        return z / (x + y)
````

````python
# Executando a função abaixo, observe que o valor de z não foi repassado, pois o mesmo foi definido com argumento nomeando.
my_function2(2, 5)
````

````python
# Mesmo tento sido nomeado, é possível definí-lo apenas considerando sua posição, sem nomeá-lo. Mas é
# ALTAMENTE recomendado sempre usar o nome

# Não recomendado
my_function2(2, 5 , 3.5)
````

````python
# Método recomendado
my_function2(2, 5 , z = 3.5)
````

## Namespaces, escopo e funções locais
Em Python, funções podem acessar variáveis definidas dentro delas (escopo local), bem como variáveis definidas fora (em escopos mais amplos, como o escopo global).

Segundo McKinney (2023), “um nome alternativo e mais descritivo para o escopo de uma variável em Python é namespace”. Embora os conceitos não sejam idênticos, eles estão intimamente relacionados: o namespace é o ambiente que armazena os nomes definidos, enquanto o escopo determina onde esses nomes podem ser acessados.

Tipos de escopo (ou namespaces comumente associados):
- Global: nomes definidos fora de qualquer função.

- Local: nomes definidos dentro de uma função.

A seguir, vejamos um exemplo de função com um namespace local:

````python
def func():
    a = []
    for i in range(5):
        a.append(i)
````

Na função acima a variável `a` só existe no momento da exução da função. Ao término, ela é destruída. 

````python
# Tente printar a variável a observe o erro
# func()
# print(a)
````

No exemplo abaixo, a variável `a` ocupa um namespace global 

````python
a = []
def func():
    for i in range(5):
        a.append(i)
````

````python
func()
print(a)
````

````python
func()
print(a)
````

### Docstring
Nas funções criadas anteriormente, observamos que a ordem dos argumentos influencia diretamente nos resultados. Outro aspecto importante é a conversão adequada das unidades de medida. O usuário precisa saber, antecipadamente, qual é a ordem correta dos argumentos e quais unidades devem ser utilizadas.

Para evitar erros e tornar as funções mais fáceis de usar, é fundamental fornecer orientações claras sobre o que cada função espera receber como entrada. Em Python, isso é feito por meio das docstrings.

Docstrings são instruções escritas logo após o cabeçalho da função (linha def), delimitadas por aspas triplas """. Elas descrevem o propósito da função, os parâmetros esperados e, opcionalmente, o que é retornado.

Veja um exemplo:

```
def escala(x, y):

    Calula o numerador da escala cartográfica dado os valores de distância gráfica (x) e distância real (y).

    Parâmetros:
    x = distância gráfica em centímetros
    y = distância real em quilômetros

    Retorna:
    O numerador da escala (distância real convertida em centímetros dividida pela distância gráfica).
    """
    y = y * 100000
    return y / x
```

### Mapeando um função sobre uma sequência de valores

````python
def quadrado(seq):
    """
    Calcula o quadrado dos valores de uma sequências de valores e cria uma 
    nova lista com os valores calculados

    Parâmetros:
    - sequências de valores (tuplas, listas,set)
    """
    lista2 = []
    for x in seq:
        x = x**2
        lista2.append(x)
    print("Lista 2:", lista2)

````

````python
# Criando uma lista de valores de 0 a 100 saltando de 5 em 5
lista = list(range(0, 101, 5))
print(lista)

quadrado(lista)
````

````python
# Criando um conjunto de valores

conjunto = {2, 5, 5, 9, 11, 5, 10, 20}

quadrado(conjunto)
````
